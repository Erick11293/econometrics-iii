```python
"""Synthetic Control Analysis of California's Proposition 99"""

from matplotlib import pyplot as plt
import pyreadr
import numpy as np
import pandas as pd
from scipy.optimize import minimize
from sklearn.metrics import mean_squared_error

# Constants
START_YEAR, INTERVENTION_YEAR, END_YEAR = 1970, 1989, 2001
EXCLUDED_STATES = ['Massachusetts', 'Arizona', 'Oregon', 'Florida', 'Alaska', 'Hawaii', 
                   'Maryland', 'Michigan', 'New Jersey', 'New York', 'Washington', 'District of Columbia']

def load_data():
    """Load and prepare outcome and predictor data"""
    # Load outcome data
    df_raw = pd.read_csv('../data/prop99.csv')
    df_outcome = pd.DataFrame(df_raw[df_raw['SubMeasureDesc'] == 'Cigarette Consumption (Pack Sales Per Capita)']
                             .pivot_table(values='Data_Value', index='LocationDesc', columns=['Year']).to_records())
    df_outcome = df_outcome[~df_outcome['LocationDesc'].isin(EXCLUDED_STATES)].reset_index()
    
    # Load predictor data
    df_predictors = pd.DataFrame(list(pyreadr.read_r('../data/smoking.rda').values())[0])
    
    return df_outcome, df_predictors

def prepare_outcome_data(df_outcome):
    """Prepare outcome data with clear pre/post intervention splits"""
    # Define year columns
    pre_years = [str(year) for year in range(START_YEAR, INTERVENTION_YEAR)]
    post_years = [str(year) for year in range(INTERVENTION_YEAR, END_YEAR)]
    all_years = pre_years + post_years
    
    # Create separate DataFrames for California and control states
    ca_outcomes = df_outcome[df_outcome['LocationDesc'] == 'California'][['LocationDesc'] + all_years].copy()
    control_outcomes = df_outcome[df_outcome['LocationDesc'] != 'California'][['LocationDesc'] + all_years].copy()
    
    # Split by time period
    ca_pre = ca_outcomes[['LocationDesc'] + pre_years].copy()
    ca_post = ca_outcomes[['LocationDesc'] + post_years].copy()
    control_pre = control_outcomes[['LocationDesc'] + pre_years].copy()
    control_post = control_outcomes[['LocationDesc'] + post_years].copy()
    
    print(f"Data shapes:")
    print(f"  California pre-intervention: {ca_pre.shape}")
    print(f"  California post-intervention: {ca_post.shape}")
    print(f"  Control states pre-intervention: {control_pre.shape}")
    print(f"  Control states post-intervention: {control_post.shape}")
    
    return {
        'ca_pre': ca_pre,
        'ca_post': ca_post, 
        'control_pre': control_pre,
        'control_post': control_post,
        'pre_years': pre_years,
        'post_years': post_years
    }

def prepare_predictor_data(df_outcome, df_predictors):
    """Prepare predictor variables with transparent data sources"""
    
    predictor_data = []
    
    for _, state_row in df_outcome.iterrows():
        state_name = state_row['LocationDesc']
        
        # Get historical cigarette sales (outcome lags as predictors)
        cigsale_1988 = state_row['1988']
        cigsale_1980 = state_row['1980'] 
        cigsale_1975 = state_row['1975']
        
        # Get state ID for predictor data matching
        ca_idx = df_outcome[df_outcome['LocationDesc'] == 'California'].index[0]
        state_id = ca_idx + 1 if state_name == 'California' else df_outcome[df_outcome['LocationDesc'] == state_name].index[0] + 1
        
        # Filter predictor data for this state
        state_predictors = df_predictors[df_predictors['state'] == state_id].copy()
        
        if len(state_predictors) == 0:
            print(f"Warning: No predictor data found for {state_name} (state_id: {state_id})")
            continue
            
        # Calculate time-averaged predictors with clear time windows
        beer_avg = state_predictors[
            (state_predictors['year'] >= 1984) & (state_predictors['year'] < INTERVENTION_YEAR)
        ]['beer'].mean()
        
        age15to24_avg = state_predictors[
            (state_predictors['year'] >= 1980) & (state_predictors['year'] < INTERVENTION_YEAR)
        ]['age15to24'].mean() * 100
        
        retprice_avg = state_predictors[
            (state_predictors['year'] >= 1980) & (state_predictors['year'] < INTERVENTION_YEAR)  
        ]['retprice'].mean()
        
        lnincome_avg = state_predictors[
            (state_predictors['year'] >= 1980) & (state_predictors['year'] < INTERVENTION_YEAR)
        ]['lnincome'].mean()
        
        predictor_data.append({
            'state': state_name,
            'lnincome_1980_1988': lnincome_avg,
            'age15to24_1980_1988': age15to24_avg, 
            'retprice_1980_1988': retprice_avg,
            'beer_1984_1988': beer_avg,
            'cigsale_1988': cigsale_1988,
            'cigsale_1980': cigsale_1980,
            'cigsale_1975': cigsale_1975
        })
    
    # Create DataFrame with predictors
    predictors_df = pd.DataFrame(predictor_data)
    
    # Split into California and control
    ca_predictors = predictors_df[predictors_df['state'] == 'California'].copy()
    control_predictors = predictors_df[predictors_df['state'] != 'California'].copy()
    
    print(f"\nPredictor variables:")
    print(f"  California predictors: {ca_predictors.shape}")
    print(f"  Control states predictors: {control_predictors.shape}")
    print(f"  Predictor columns: {list(predictors_df.columns[1:])}")
    
    return {
        'ca_predictors': ca_predictors,
        'control_predictors': control_predictors,
        'predictor_columns': list(predictors_df.columns[1:])
    }

def convert_to_matrices(outcome_data, predictor_data):
    """Convert DataFrames to matrices for optimization (final step)"""
    
    # Extract year columns (excluding 'LocationDesc')
    pre_year_cols = outcome_data['pre_years']
    post_year_cols = outcome_data['post_years']
    predictor_cols = predictor_data['predictor_columns']
    
    print(f"\nConverting to matrices for optimization:")
    print(f"  Using {len(pre_year_cols)} pre-intervention years: {pre_year_cols[0]}-{pre_year_cols[-1]}")
    print(f"  Using {len(post_year_cols)} post-intervention years: {post_year_cols[0]}-{post_year_cols[-1]}")
    print(f"  Using {len(predictor_cols)} predictor variables")
    
    # Outcome matrices
    Z1 = outcome_data['ca_pre'][pre_year_cols].values.T  # CA pre-intervention (years x 1)
    Y1 = outcome_data['ca_post'][post_year_cols].values.T  # CA post-intervention (years x 1)
    Z0 = outcome_data['control_pre'][pre_year_cols].values.T  # Control pre-intervention (years x states)
    Y0 = outcome_data['control_post'][post_year_cols].values.T  # Control post-intervention (years x states)
    
    # Predictor matrices  
    X1 = predictor_data['ca_predictors'][predictor_cols].values.T  # CA predictors (predictors x 1)
    X0 = predictor_data['control_predictors'][predictor_cols].values.T  # Control predictors (predictors x states)
    
    print(f"  Matrix shapes - Z1: {Z1.shape}, Z0: {Z0.shape}, Y1: {Y1.shape}, Y0: {Y0.shape}")
    print(f"  Matrix shapes - X1: {X1.shape}, X0: {X0.shape}")
    
    return Z0, Z1, Y0, Y1, X0, X1

def solve_synthetic_control(Z0, Z1, control_states):
    """Solve synthetic control using simple optimization"""
    n_donors = Z0.shape[1]
    
    def objective(weights):
        weights = weights / weights.sum()  # Normalize to sum to 1
        synthetic = Z0.dot(weights).reshape(-1, 1)
        return mean_squared_error(Z1, synthetic)
    
    # Constraints and bounds
    constraints = {'type': 'eq', 'fun': lambda w: w.sum() - 1}
    bounds = [(0, 1) for _ in range(n_donors)]
    w0 = np.ones(n_donors) / n_donors
    
    result = minimize(objective, w0, method='SLSQP', bounds=bounds, constraints=constraints)
    
    # Create results DataFrame showing donor weights
    weights_df = pd.DataFrame({
        'state': control_states,
        'weight': result.x
    }).sort_values('weight', ascending=False)
    
    print(f"\nSynthetic Control Weights (top 10):")
    print(weights_df.head(10).to_string(index=False, float_format='%.4f'))
    print(f"Sum of weights: {result.x.sum():.6f}")
    
    return result.x, weights_df

def plot_results(outcome_data, weights, weights_df):
    """Generate plots using DataFrames for transparency"""
    
    # Reconstruct full time series using DataFrames
    all_years = outcome_data['pre_years'] + outcome_data['post_years']
    years = [int(year) for year in all_years]
    
    # California outcomes
    ca_full = pd.concat([
        outcome_data['ca_pre'][outcome_data['pre_years']], 
        outcome_data['ca_post'][outcome_data['post_years']]
    ], axis=1)
    ca_values = ca_full.iloc[0, :].values
    
    # Control outcomes  
    control_full = pd.concat([
        outcome_data['control_pre'][outcome_data['pre_years']],
        outcome_data['control_post'][outcome_data['post_years']]
    ], axis=1)
    
    # Calculate synthetic California using weights
    synthetic_values = control_full.values.T.dot(weights)
    
    # Create plots
    plt.figure(figsize=(15, 5))
    
    # Plot 1: Trends comparison
    plt.subplot(1, 3, 1)
    plt.plot(years, synthetic_values, 'r--', label="Synthetic California", linewidth=2)
    plt.plot(years, ca_values, 'b-', label="California", linewidth=2)
    plt.axvline(INTERVENTION_YEAR, color='black', alpha=0.7)
    plt.text(INTERVENTION_YEAR + 0.2, 30, 'Prop 99', rotation=90)
    plt.ylabel('Cigarette sales (packs per capita)')
    plt.xlabel('Year')
    plt.title('California vs. Synthetic California')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot 2: Treatment effect
    gap = synthetic_values - ca_values
    plt.subplot(1, 3, 2)
    plt.plot(years, gap, 'purple', linewidth=2)
    plt.axhline(0, color='black', alpha=0.5)
    plt.axvline(INTERVENTION_YEAR, color='black', linestyle='--', alpha=0.7)
    plt.ylabel('Treatment Effect (packs per capita)')
    plt.xlabel('Year')
    plt.title('Gap: Synthetic CA - Actual CA')
    plt.grid(True, alpha=0.3)
    
    # Plot 3: Donor state weights
    plt.subplot(1, 3, 3)
    top_donors = weights_df.head(8)
    plt.barh(range(len(top_donors)), top_donors['weight'])
    plt.yticks(range(len(top_donors)), top_donors['state'])
    plt.xlabel('Weight')
    plt.title('Top Donor State Weights')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("prop99_results.png", dpi=300, bbox_inches='tight')
    plt.show()
    
    # Calculate and display treatment effects
    post_intervention_idx = len(outcome_data['pre_years'])
    post_gap = gap[post_intervention_idx:]
    
    print(f"\nTreatment Effects:")
    print(f"  Average post-intervention effect: {np.mean(post_gap):.2f} packs per capita")
    print(f"  Effect in final year ({years[-1]}): {gap[-1]:.2f} packs per capita")
```

```python

# Main execution
df_outcome, df_predictors = load_data()
outcome_data = prepare_outcome_data(df_outcome)
predictor_data = prepare_predictor_data(df_outcome, df_predictors)
Z0, Z1, Y0, Y1, X0, X1 = convert_to_matrices(outcome_data, predictor_data)

# Get control state names for results display
control_states = outcome_data['control_pre']['LocationDesc'].tolist()
weights, weights_df = solve_synthetic_control(Z0, Z1, control_states)
plot_results(outcome_data, weights, weights_df)
```
